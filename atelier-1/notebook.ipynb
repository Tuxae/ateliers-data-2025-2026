```python
X = df.drop(columns=['malignant'])
y = df['malignant']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=12345, stratify=y) # stratification
```

```python
y.value_counts(normalize=True), y_train.value_counts(normalize=True), y_test.value_counts(normalize=True)
```

```python
# Fusion de X_train et y_train
df_train = X_train.copy()
df_train['malignant'] = y_train

# Préparation de X_test avec '?' pour "malevolent"
df_test = X_test.copy()
df_test['malignant'] = '?'

# Concaténation finale
pd.concat([df_train, df_test], ignore_index=True).plot(kind='scatter', x='texture_mean', y='smoothness_mean', color='malignant')
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(X_train, y_train)
```

```python
sum(knn.predict(X_test) == y_test)/len(X_test) #knn.score(X_test, y_test)
```

```python
knn.score(X_test, y_test)
```

```python
scores = pd.DataFrame(data=[i for i in range(2, 300)], columns=['score'], index=range(2, 300))

scores['score'] = scores['score'].apply(lambda i: KNeighborsClassifier(n_neighbors=i).fit(X_train, y_train).score(X_test, y_test))
scores.plot()
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_scores = cross_val_score(knn, X, y, cv=skf)

print("Scores CV:", cv_scores)
print("Score CV min:", cv_scores.min())
```

```python
scores = pd.DataFrame(data=[i for i in range(2, 300)], columns=['score'], index=range(2, 300))

scores['score'] = scores['score'].apply(lambda i: cross_val_score(KNeighborsClassifier(n_neighbors=i), X, y, cv=skf).min())
scores.plot()
```

```python
prediction = knn.predict(X_test)

VP = sum(prediction[y_test == True])
FN = sum(~prediction[y_test == True])
FP = sum(~y_test[prediction == True])

precision = VP/(VP + FP)
recall = VP/(VP + FN)
f1 = 2*precision*recall/(precision+recall)

precision, recall, f1
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

precision_score(y_test, prediction), recall_score(y_test, prediction), f1_score(y_test, prediction)
```

```python
cv_scores = cross_val_score(knn, X, y, cv=skf, scoring='f1')

print("Scores CV:", cv_scores)
print("Score CV min:", cv_scores.min())
```

```python
scores = pd.DataFrame(data=[i for i in range(2, 300)], columns=['score'], index=range(2, 300))

scores['score_acc'] = scores['score'].apply(lambda i: cross_val_score(KNeighborsClassifier(n_neighbors=i), X, y, cv=skf).min())
scores['score_f1'] = scores['score'].apply(lambda i: cross_val_score(KNeighborsClassifier(n_neighbors=i), X, y, cv=skf, scoring='f1').min())
scores[['score_acc', 'score_f1']].plot()
```

```python
knn = KNeighborsClassifier(n_neighbors=17)

cross_val_score(knn, X, y, cv=skf, scoring='f1').min()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, RobustScaler

pipe = Pipeline([
    ('scale', StandardScaler()),
    ('clf', knn)
    ])

cross_val_score(pipe, X, y, cv=skf, scoring='f1').min()
```

```python
from sklearn.preprocessing import RobustScaler

pipe = Pipeline([
    ('scale', RobustScaler()),
    ('clf', knn)
    ])

cross_val_score(pipe, X, y, cv=skf, scoring='f1').min()
```

```python
coef_df_sorted = coef_df.reindex(coef_df['coefficient'].abs().sort_values(ascending=False).index)
coef_df_sorted.plot(x='feature', y='coefficient', kind='bar')
```

```python
top_n_features = coef_df_sorted.head(15)["feature"].values
top_n_features
```

```python
cross_val_score(pipe, X, y, cv=skf, scoring='f1').min(), cross_val_score(pipe, X[top_n_features], y, cv=skf, scoring='f1').min()
```

```python
from sklearn.inspection import permutation_importance

pipe.fit(X_train, y_train)
result = permutation_importance(
    pipe, X_test, y_test, n_repeats=10, random_state=42, n_jobs=2
)

sorted_importances_idx = result.importances_mean.argsort()
importances = pd.DataFrame(
    result.importances[sorted_importances_idx].T,
    columns=X.columns[sorted_importances_idx],
)
importances.plot(kind='box')
```
