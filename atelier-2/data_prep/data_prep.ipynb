```python
import pandas as pd
auto_mpg = pd.read_csv('auto-mpg.xls', sep = ',')
auto_mpg
```

```python
df = auto_mpg[['car name', 'cylinders', 'displacement', 'horsepower', 'weight',
       'acceleration', 'model year', 'origin', 'mpg']]
```

I'll randomly replace a small fraction of values (5%) in the dataframe with NaN to introduce some missing data.

```python
import numpy as np

drop_frac = 0.02
df_with_nan = df.copy()

for col in df_with_nan.columns:
    if col != 'car name' and col != 'mpg':
        mask = np.random.rand(len(df_with_nan)) < drop_frac
        df_with_nan.loc[mask, col] = np.nan

df_with_nan
```

```python
df_with_nan.to_csv("dataset_car.csv", index=False)
```
