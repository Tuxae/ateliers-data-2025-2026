```python
import pandas as pd
import numpy as np

pd.options.plotting.backend = "plotly"
```

```python
df_car[df_car['horsepower'].str.isnumeric() != True] #important, si on met == False
```

```python
df_car=df_car.replace('?',np.nan)
df_car[df_car['horsepower'].str.isnumeric() != True]
```

```python
df_car.info()
```

```python
df_car['horsepower'] = df_car['horsepower'].astype(np.float64)
```

```python
df_car.info()
```

```python
df_car.isna().plot(kind='imshow')
```

```python
X = df_car.drop(columns=['car name', 'mpg'])
y = df_car['mpg']
```

```python
X.isna().plot(kind='imshow')
```

```python
X.apply(lambda x: x.fillna(x.median()),axis=0).isna().plot(kind='imshow')
```

```python
from sklearn.impute import SimpleImputer

median_imputer = SimpleImputer(strategy='median')
X_imputed_median = pd.DataFrame(
    median_imputer.fit_transform(X),
    columns=X.columns
)
X_imputed_median.isna().plot(kind='imshow')
X_imputed_median
```

```python
from sklearn.impute import KNNImputer

knn_imputer = KNNImputer(n_neighbors=5)
X_imputed_knn = pd.DataFrame(
    knn_imputer.fit_transform(X),
    columns=X.columns
)
X_imputed_knn.isna().plot(kind='imshow')
X_imputed_knn
```

```python
df = X_imputed_knn.join(y)
```

```python
df.plot(kind='scatter', x='weight', y='mpg')
```

```python
df.plot(kind='scatter', x='horsepower', y='mpg')
```
