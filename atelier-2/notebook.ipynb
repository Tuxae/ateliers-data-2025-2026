```python
X = df_car.drop(columns=['car name', 'mpg'])
y = df_car['mpg']
```

```python
X.isna().plot(kind='imshow')
```

```python
X.apply(lambda x: x.fillna(x.median()),axis=0).isna().plot(kind='imshow')
```

```python
from sklearn.impute import SimpleImputer

median_imputer = SimpleImputer(strategy='median')
X_imputed_median = pd.DataFrame(
    median_imputer.fit_transform(X),
    columns=X.columns
)
X_imputed_median.isna().plot(kind='imshow')
X_imputed_median
```

```python
from sklearn.impute import KNNImputer

knn_imputer = KNNImputer(n_neighbors=5)
X_imputed_knn = pd.DataFrame(
    knn_imputer.fit_transform(X),
    columns=X.columns
)
X_imputed_knn.isna().plot(kind='imshow')
X_imputed_knn
```

```python
df = X_imputed_knn.join(y)
```

```python
df.plot(kind='scatter', x='weight', y='mpg')
```

```python
df.plot(kind='scatter', x='horsepower', y='mpg')
```

```python
df.plot(kind='scatter', x='model year', y='mpg')
```

```python
from sklearn.linear_model import LinearRegression
import plotly.graph_objects as go

X_weight = df[['weight']]
y_mpg = df['mpg']

model = LinearRegression()
model.fit(X_weight, y_mpg)

y_pred = model.predict(X_weight)

fig = go.Figure()
fig.add_trace(go.Scatter(x=df['weight'], y=df['mpg'], mode='markers', name='Data', showlegend=False))
fig.add_trace(go.Scatter(x=df['weight'], y=y_pred, mode='lines', name='Linear Regression', showlegend=False))
fig.update_xaxes(title_text='weight')
fig.update_yaxes(title_text='mpg')
fig.show()
```

```python
from sklearn.linear_model import LinearRegression
import plotly.graph_objects as go

X_3d = df[['weight', 'model year']]
y_mpg = df['mpg']

model_3d = LinearRegression()
model_3d.fit(X_3d, y_mpg)

weight_range = np.linspace(df['weight'].min(), df['weight'].max(), 20)
model_year_range = np.linspace(df['model year'].min(), df['model year'].max(), 20)
weight_grid, model_year_grid = np.meshgrid(weight_range, model_year_range)

X_grid = np.c_[weight_grid.ravel(), model_year_grid.ravel()]
y_pred_grid = model_3d.predict(X_grid).reshape(weight_grid.shape)

fig = go.Figure()

fig.add_trace(go.Scatter3d(
    x=df['weight'],
    y=df['model year'],
    z=df['mpg'],
    mode='markers',
    marker=dict(size=3, color='blue'),
    name='Data'
))

fig.add_trace(go.Surface(
    x=weight_grid,
    y=model_year_grid,
    z=y_pred_grid,
    colorscale='Viridis',
    opacity=0.7,
    name='Regression Plane'
))

fig.update_layout(
    scene=dict(
        xaxis_title='Weight',
        yaxis_title='Model Year',
        zaxis_title='MPG'
    ),
    height=800,
)

fig.show()
```

```python
X_imputed_knn[["model year", "weight"]]
```

```python
from sklearn.preprocessing import add_dummy_feature

X_b = add_dummy_feature(X_imputed_knn[["model year", "weight"]])
X_b
```

```python
X_b = X_imputed_knn[["model year", "weight"]].copy()
X_b['intercept'] = 1
X_b
```

```python
theta_ols = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
theta_ols
```

```python
y_pred_ols = X_b.values @ theta_ols.values

weight_range = np.linspace(df['weight'].min(), df['weight'].max(), 20)
model_year_range = np.linspace(df['model year'].min(), df['model year'].max(), 20)
weight_grid, model_year_grid = np.meshgrid(weight_range, model_year_range)

X_grid = np.c_[model_year_grid.ravel(), weight_grid.ravel(), np.ones(weight_grid.size)]
y_pred_grid = (X_grid @ theta_ols).reshape(weight_grid.shape)

fig = go.Figure([
    go.Scatter3d(x=df['weight'], y=df['model year'], z=df['mpg'], mode='markers', marker=dict(size=3)),
    go.Surface(x=weight_grid, y=model_year_grid, z=y_pred_grid, opacity=0.7)
])

fig.update_layout(scene=dict(xaxis_title='Weight', yaxis_title='Model Year', zaxis_title='MPG'), height=800)
fig.show()
```

```python
theta_ols = np.linalg.pinv(X_b) @ y
theta_ols
```

```python
y_pred_ols = X_b.values @ theta_ols

weight_range = np.linspace(df['weight'].min(), df['weight'].max(), 20)
model_year_range = np.linspace(df['model year'].min(), df['model year'].max(), 20)
weight_grid, model_year_grid = np.meshgrid(weight_range, model_year_range)

X_grid = np.c_[model_year_grid.ravel(), weight_grid.ravel(), np.ones(weight_grid.size)]
y_pred_grid = (X_grid @ theta_ols).reshape(weight_grid.shape)

fig = go.Figure([
    go.Scatter3d(x=df['weight'], y=df['model year'], z=df['mpg'], mode='markers', marker=dict(size=3)),
    go.Surface(x=weight_grid, y=model_year_grid, z=y_pred_grid, opacity=0.7)
])

fig.update_layout(scene=dict(xaxis_title='Weight', yaxis_title='Model Year', zaxis_title='MPG'), height=800)
fig.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

X_train, X_test, y_train, y_test = train_test_split(X_imputed_knn, y, test_size=0.2, random_state=42)

lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)

y_pred_train = lin_reg.predict(X_train)
y_pred_test = lin_reg.predict(X_test)

print(f"Training R² Score: {r2_score(y_train, y_pred_train):.4f}")
print(f"Test R² Score: {r2_score(y_test, y_pred_test):.4f}")
print(f"Training RMSE: {np.sqrt(mean_squared_error(y_train, y_pred_train)):.4f}")
print(f"Test RMSE: {np.sqrt(mean_squared_error(y_test, y_pred_test)):.4f}")
```

```python
lin_reg.fit(X_imputed_knn, y)
coefs = pd.DataFrame({
    'feature': X_imputed_knn.columns,
    'coef': lin_reg.coef_,
    'model': 'linear'
})
coefs.plot(kind='bar', y='coef', x='feature')
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=y_test, y=y_pred_test, mode='markers', name='Test Data'))
fig.add_trace(go.Scatter(x=[y_test.min(), y_test.max()], y=[y_test.min(), y_test.max()],
                         mode='lines', name='Perfect Prediction', line=dict(dash='dash')))
fig.update_xaxes(title_text='Actual MPG')
fig.update_yaxes(title_text='Predicted MPG')
fig.update_layout(title='Actual vs Predicted MPG')
fig.show()
```

```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet

X_train_full = X_imputed_knn
y_train_full = y

ridge = Ridge(alpha=1.0)
ridge.fit(X_train_full, y_train_full)

y_pred_ridge_train = ridge.predict(X_train)
y_pred_ridge_test = ridge.predict(X_test)

print("Ridge Regression:")
print(f"Training R² Score: {r2_score(y_train, y_pred_ridge_train):.4f}")
print(f"Test R² Score: {r2_score(y_test, y_pred_ridge_test):.4f}")
print(f"Training RMSE: {np.sqrt(mean_squared_error(y_train, y_pred_ridge_train)):.4f}")
print(f"Test RMSE: {np.sqrt(mean_squared_error(y_test, y_pred_ridge_test)):.4f}")
```

```python
from sklearn.model_selection import cross_val_score

alphas = np.logspace(-8, 3, 300)
test_rmse = []

for alpha in alphas:
    ridge_model = Ridge(alpha=alpha)
    cv_scores = cross_val_score(ridge_model, X_imputed_knn, y, cv=5,
                                scoring='neg_root_mean_squared_error')
    test_rmse.append(-cv_scores.mean())
    
fig = go.Figure()
fig.add_trace(go.Scatter(x=alphas, y=test_rmse, mode='lines+markers'))
fig.update_xaxes(title_text='Alpha', type='log')
fig.update_yaxes(title_text='Test RMSE (CV)')
fig.show()
```

```python
best_alpha = alphas[np.argmin(test_rmse)]
best_rmse = min(test_rmse)
print(f"Best Alpha: {best_alpha:.4f}")
print(f"Best CV RMSE: {best_rmse:.4f}")
```

```python
ridge = Ridge(alpha=best_alpha)
ridge.fit(X_imputed_knn, y)
coefs_ridge = pd.DataFrame({
    'feature': X_imputed_knn.columns,
    'coef': ridge.coef_,
    'model': 'ridge'
})

coefs = pd.concat([coefs_ridge, coefs], ignore_index=True)
coefs.plot(kind='bar', y='coef', x='feature', color='model', barmode="group")
```
