```python
from numpy.typing import NDArray

class Perceptron:
    w: NDArray
    b: float
    lr: float

    def __init__(self, learning_rate: float):
        self.w = np.zeros(28 * 28)
        self.b = 0.0
        self.lr = learning_rate

    def forward(self, x_i: NDArray) -> int:
        act = self.w @ x_i + self.b
        return 1 if act > 0 else 0

    def update(self, x_i: NDArray, y_i: int) -> None:
        y_i_hat = self.forward(x_i)
        self.w += self.lr * (y_i - y_i_hat) * x_i
```

```python
def train_loss_perceptron(p: Perceptron) -> float:
    loss = 0.0
    n = X_train.shape[0]
    for i in range(n):
        x_i = X_train.iloc[i].to_numpy()
        y_i = y_train.iloc[i]
        y_i_hat = p.forward(x_i)
        loss += np.abs(y_i - y_i_hat)
    return loss / n

def test_loss_perceptron(p: Perceptron) -> float:
    loss = 0.0
    n = X_test.shape[0]
    for i in range(n):
        x_i = X_test.iloc[i].to_numpy()
        y_i = y_test.iloc[i]
        y_i_hat = p.forward(x_i)
        loss += np.abs(y_i - y_i_hat)
    return loss / n
```

```python
def train_perceptron(p: Perceptron, max_epochs: int, stop_under = 0.001) -> pd.DataFrame:
    metrics = []
    n = X_train.shape[0]

    for epoch in range(max_epochs):
        for i in range(n):
            x_i = X_train.iloc[i].to_numpy()
            y_i = y_train.iloc[i]
            p.update(x_i, y_i)

        train_loss = train_loss_perceptron(p)
        test_loss = test_loss_perceptron(p)
        metrics.append([epoch, train_loss, test_loss])
        print(f"End of epoch {epoch}: Train loss {train_loss:.3f}, Test loss {test_loss:.3f}")
        if train_loss < stop_under:
            print(f"Train loss {train_loss:.3f} < {stop_under:.3f}, stopping. Test loss {test_loss:.3f}")
            break

    return pd.DataFrame(metrics, columns=["epoch", "train_loss", "test_loss"]).set_index("epoch")
```

```python
from numpy.typing import NDArray

class LayerPerceptron:
    dim: int
    w: NDArray
    b: NDArray
    lr: float

    def __init__(self, dim: int, learning_rate: float):
        self.dim = dim
        self.w = np.zeros((dim, 28 * 28))
        self.b = np.zeros(dim)
        self.lr = learning_rate

    def forward(self, x_i: NDArray) -> NDArray:
        act = self.w @ x_i + self.b
        return (act > 0).astype(int)

    def update(self, x_i: NDArray, y_i: NDArray) -> None:
        y_i_hat = self.forward(x_i)
        self.w += self.lr * (y_i - y_i_hat).reshape(self.dim, 1) @ x_i.reshape(1, 28*28)
```

```python
def train_loss_layer_perceptron(p: LayerPerceptron) -> float:
    loss = 0.0
    n = X_train.shape[0]
    for i in range(n):
        x_i = X_train.iloc[i].to_numpy()
        y_i = y_train.iloc[i].to_numpy()
        y_i_hat = p.forward(x_i)
        loss += np.abs(y_i - y_i_hat).sum() #Loss L1
    return loss / n

def test_loss_layer_perceptron(p: LayerPerceptron) -> float:
    loss = 0.0
    n = X_test.shape[0]
    for i in range(n):
        x_i = X_test.iloc[i].to_numpy()
        y_i = y_test.iloc[i].to_numpy()
        y_i_hat = p.forward(x_i)
        loss += np.abs(y_i - y_i_hat).sum()
    return loss / n
```

```python
layer_perceptron = LayerPerceptron(2, 0.001)
train_loss_layer_perceptron(layer_perceptron)
```

```python
def train_layer_perceptron(p: LayerPerceptron, max_epochs: int, stop_under = 0.001) -> pd.DataFrame:
    metrics = []
    n = X_train.shape[0]

    for epoch in range(max_epochs):
        for i in range(n):
            x_i = X_train.iloc[i].to_numpy()
            y_i = y_train.iloc[i].to_numpy()
            p.update(x_i, y_i)

        train_loss = train_loss_layer_perceptron(p)
        test_loss = test_loss_layer_perceptron(p)
        metrics.append([epoch, train_loss, test_loss])
        print(f"End of epoch {epoch}: Train loss {train_loss:.3f}, Test loss {test_loss:.3f}")
        if train_loss < stop_under:
            print(f"Train loss {train_loss:.3f} < {stop_under:.3f}, stopping. Test loss {test_loss:.3f}")
            break

    return pd.DataFrame(metrics, columns=["epoch", "train_loss", "test_loss"]).set_index("epoch")
```

```python
layer_perceptron = LayerPerceptron(2, 0.001)
metrics = train_layer_perceptron(layer_perceptron, 500)
metrics.plot()
```

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

X_train = X_train.to_numpy().astype(np.float32)
X_test = X_test.to_numpy().astype(np.float32)
```

```python
cats = pd.Categorical(pd.concat([y_train, y_test], axis=0))
y_train = pd.Categorical(y_train, categories=cats.categories).codes.astype(np.int64)
y_test = pd.Categorical(y_test, categories=cats.categories).codes.astype(np.int64)

n_features = X_train.shape[1]
n_classes = int(len(cats.categories))
```

```python
train_ds = TensorDataset(torch.from_numpy(X_train), torch.from_numpy(y_train))
test_ds = TensorDataset(torch.from_numpy(X_test), torch.from_numpy(y_test))
train_loader = DataLoader(train_ds, batch_size=20, shuffle=True)
test_loader = DataLoader(test_ds)
```

```python
model = nn.Sequential(
    nn.Linear(n_features, 64),
    nn.ReLU(),
    nn.Linear(64, n_classes),
)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=1e-3)
```

```python
def avg_loss(loader: DataLoader) -> float:
    model.eval()
    running_loss = 0.0
    n_seen = 0
    with torch.no_grad():
        for xb, yb in loader:
            xb = xb
            yb = yb

            logits = model(xb)
            loss = criterion(logits, yb)
            running_loss += loss.item() * xb.size(0)
            n_seen += xb.size(0)
    return running_loss / max(n_seen, 1)
```

```python
metrics = []

for epoch in range(10_000):
    model.train()
    running_loss = 0.0
    n_seen = 0

    for xb, yb in train_loader:
        xb = xb
        yb = yb

        optimizer.zero_grad()
        logits = model(xb)
        loss = criterion(logits, yb)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * xb.size(0)
        n_seen += xb.size(0)

    train_loss = running_loss / max(n_seen, 1)
    test_loss = avg_loss(test_loader)

    metrics.append([epoch, train_loss, test_loss])

    print(f"End of epoch {epoch}: Train loss {train_loss:.4f}, Test loss {test_loss:.4f}")

pd.DataFrame(metrics, columns=["epoch", "train_loss", "test_loss"]).set_index("epoch").plot()
```
